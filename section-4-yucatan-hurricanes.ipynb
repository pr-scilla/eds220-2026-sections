## 1. Archive Exploration

Dataset description: 
This is a public dataset. For each hurricane recorded in the dataset, there is:
- a unique ID code
- start date
- end date
- tropical storm number
- hurricane name
- Saffir-Simpson category at landfall or closest approach to the Yucatan Peninsula
- reconstructed maximum damage on the Fujita scale (F0-F3) across the Yucatan Peninsula. Nd = no damage.

Citation: 
Boose, E. and D. Foster. 2023. Ecological Impacts of Hurricanes Across the Yucatan Peninsula 1851-2000 ver 23. Environmental Data Initiative. https://doi.org/10.6073/pasta/f219113373913f2daf421732e28d3c38 (Accessed 2026-10-22).

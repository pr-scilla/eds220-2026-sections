# Week 1 Discussion Section: Prey species in the California drylands

## About the data

Rachel King, Jenna Braun, Michael Westphal, & CJ Lortie. (2023). *Compiled occurrence records for prey items of listed species found in California drylands with associated environmental data*. Knowledge Network for Biocomplexity. [doi:10.5063/F1VM49RH](https://knb.ecoinformatics.org/view/doi:10.5063/F1VM49RH).

Date accessed: 10/1/2026

This dataset contains occurence records for arthropod prey of endangered vertebrate species in California drylands. The dataset also contains climate and environmental data associated with the invertebrate prey records.

- The data were not collected by the authors in-situ. Arthropod prey occurence data were downloaded from the Global Biodiversity Information Facility (GBIF) data resource. Climate and environmental data were compiled from other publications as referenced in the [Sampling Description](https://knb.ecoinformatics.org/view/doi%3A10.5063%2FF1VM49RH) section.
- The date range for this dataset is 1945 to 2022.
- There is a metadata file associated with this dataset.
- The data is marked as non-sensitive data.
